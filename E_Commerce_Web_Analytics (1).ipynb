```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

correlation = df_clean[
    ['TotalPages',
     'TotalDuration',
     'ProductRelated',
     'ProductRelated_Duration',
     'BounceRates',
     'ExitRates',
     'PageValues']
].corr()

print(correlation)
```

```python
from scipy.stats import pearsonr

r, p = pearsonr(
    df_clean['TotalPages'],
    df_clean['TotalDuration']
)

print("r =", r)
print("p =", p)
```

```python
from scipy.stats import chi2_contingency

table = pd.crosstab(
    df['VisitorType'],
    df['Revenue']
)

print(table)

chi2, p, dof, expected = chi2_contingency(table)

print("Chi-square =", chi2)
print("p-value =", p)
print("Degrees of freedom =", dof)
```

```python
visitor_conversion = df.groupby('VisitorType')['Revenue'].mean() * 100

print(visitor_conversion)
```

```python
group_means = df_clean.groupby('Revenue')[
    ['TotalPages',
     'TotalDuration',
     'ProductRelated',
     'ProductRelated_Duration',
     'BounceRates',
     'ExitRates',
     'PageValues']
].mean()

print(group_means)
```

```python
from scipy.stats import mannwhitneyu

variables = [
    'TotalPages',
    'TotalDuration',
    'ProductRelated',
    'ProductRelated_Duration',
    'BounceRates',
    'ExitRates',
    'PageValues'
]

for variable in variables:

    revenue = df_clean[df_clean['Revenue'] == True][variable]
    non_revenue = df_clean[df_clean['Revenue'] == False][variable]

    stat, p = mannwhitneyu(
        revenue,
        non_revenue,
        alternative='two-sided'
    )

    print(variable)
    print("U statistic:", stat)
    print("p-value:", p)
    print("-" * 40)
```
